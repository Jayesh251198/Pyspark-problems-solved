## Employees Earning More Than Their Managers

## Problem

**Difficulty:** Easy

**Topics**

- SQL

---


## Table: `Employee`

| Column Name | Type |
|-------------|------|
| id | int |
| name | varchar |
| salary | int |
| managerId | int |

- `id` is the primary key (column with unique values) for this table.
- Each row of this table indicates the ID of an employee, their name, salary, and the ID of their manager.

---

## Task

Write a solution to find the employees who earn more than their managers.

- Return the result table in **any order**.

The result format is shown in the following example.

---

## Example 1

### Input

### Employee

| id | name | salary | managerId |
|----|------|--------|-----------|
| 1 | Joe | 70000 | 3 |
| 2 | Henry | 80000 | 4 |
| 3 | Sam | 60000 | NULL |
| 4 | Max | 90000 | NULL |

---

### Output

| Employee |
|----------|
| Joe |

---

### Explanation

- **Joe** has a salary of **70000**.
- Joe's manager is **Sam** (`managerId = 3`).
- Sam's salary is **60000**.
- Since **70000 > 60000**, Joe earns more than his manager.
- Henry's manager is Max, whose salary (**90000**) is greater than Henry's salary (**80000**), so Henry is not included in the result.
- Sam and Max do not have managers (`managerId = NULL`), so they are not considered.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

# Start Spark session
spark = SparkSession.builder.appName("Databricks-PySpark-DF").getOrCreate()
```

### 🔹 PySpark Table DataFrame

```python
# Define schema
employee_schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("salary", IntegerType(), True),
    StructField("managerId", IntegerType(), True)
])

# Create data
employee_data = [
    (1, "Joe", 70000, 3),
    (2, "Henry", 80000, 4),
    (3, "Sam", 60000, None),
    (4, "Max", 90000, None)
]

# Create DataFrame
employee_df = spark.createDataFrame(employee_data, schema=employee_schema)

print("Employee Table:")
display(employee_df)
```

### 🔹 Logic Section

```python
employees_with_higher_salary_df = (employee_df.alias("e1")  
                                .join(
                                    employee_df.alias("e2"), 
                                    col("e1.id") == col("e2.managerId"),
                                    "inner"
                                )
                                .filter(col("e1.salary") < col("e2.salary"))
                                .select(col("e2.name").alias("Employee")))

display(employees_with_higher_salary_df)
```
