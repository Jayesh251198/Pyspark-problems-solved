## Customers Who Never Order

## Problem

**Difficulty:** Easy

**Topics**

- SQL

---

## Table: `Customers`

| Column Name | Type |
|-------------|------|
| id | int |
| name | varchar |

- `id` is the primary key (column with unique values) for this table.
- Each row of this table indicates the ID and name of a customer.

---

## Table: `Orders`

| Column Name | Type |
|-------------|------|
| id | int |
| customerId | int |

- `id` is the primary key (column with unique values) for this table.
- `customerId` is a foreign key (reference column) of the `id` from the `Customers` table.
- Each row of this table indicates the ID of an order and the ID of the customer who ordered it.

---

## Task

Write a solution to find all customers who never order anything.

- Return the result table in **any order**.

The result format is shown in the following example.

---

## Example 1

### Input

### Customers

| id | name |
|----|-------|
| 1 | Joe |
| 2 | Henry |
| 3 | Sam |
| 4 | Max |

### Orders

| id | customerId |
|----|------------|
| 1 | 3 |
| 2 | 1 |

---

### Output

| Customers |
|-----------|
| Henry |
| Max |

---

### Explanation

- **Joe** (`id = 1`) has an order in the `Orders` table, so he is **not** included.
- **Sam** (`id = 3`) also has an order, so he is **not** included.
- **Henry** (`id = 2`) does **not** have any matching order in the `Orders` table, so he is included in the result.
- **Max** (`id = 4`) also has **no orders**, so he is included in the result.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

# Start Spark session
spark = SparkSession.builder.appName("Databricks-PySpark-DF").getOrCreate()
```

### 🔹 PySpark Table DataFrame


```python

# Input data
customers_data = [
    (1, "Joe"),
    (2, "Henry"),
    (3, "Sam"),
    (4, "Max")
]

orders_data = [
    (1, 3),
    (2, 1)
]

# Define schemas
customers_columns = ["id", "name"]
orders_columns = ["id", "customerId"]

# Create DataFrames
customers_df = spark.createDataFrame(customers_data, customers_columns)
orders_df = spark.createDataFrame(orders_data, orders_columns)

print("customers Table:")
display(customers_df)

print("orders Table:")
display(orders_df)
```

### 🔹 Logic Section

```python
customers_who_never_order = (customers_df.alias("c")
                             .join(orders_df.alias("o"), col("c.id") == col("o.customerId"), "left")
                             .filter(col("customerId").isNull())
).select(
    col("name").alias("Customer")
    )
display(customers_who_never_order)
```

```python
customers_who_never_order = (customers_df.alias("c")
                             .join(orders_df.alias("o"), col("c.id") == col("o.customerId"), 'left_anti')
                            ).select( col("name").alias("Customer"))
display(customers_who_never_order)
```
