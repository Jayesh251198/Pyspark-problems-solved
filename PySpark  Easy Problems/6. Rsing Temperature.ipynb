## Table: `Weather`

| Column Name | Type |
|-------------|------|
| id | int |
| recordDate | date |
| temperature | int |

- `id` is the column with unique values for this table.
- There are no different rows with the same `recordDate`.
- This table contains information about the temperature on a certain day.

---

## Task

Write a solution to find all dates' `id` with **higher temperatures compared to the previous date (yesterday)**.

- Return the result table in **any order**.

The result format is shown in the following example.

---

## Example 1

### Input

### Weather

| id | recordDate | temperature |
|----|------------|-------------|
| 1 | 2015-01-01 | 10 |
| 2 | 2015-01-02 | 25 |
| 3 | 2015-01-03 | 20 |
| 4 | 2015-01-04 | 30 |

---

### Output

| id |
|----|
| 2 |
| 4 |

---

### Explanation

- On **2015-01-01**, there is no previous day available for comparison.
- On **2015-01-02**, the temperature increased from **10** to **25**, so **`id = 2`** is included.
- On **2015-01-03**, the temperature decreased from **25** to **20**, so **`id = 3`** is **not** included.
- On **2015-01-04**, the temperature increased from **20** to **30**, so **`id = 4`** is included.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

# Start Spark session
spark = SparkSession.builder.appName("Databricks-PySpark-DF").getOrCreate()
```

### 🔹 PySpark Table DataFrame


```python
# Input data
weather_data = [
    (1, "2015-01-01", 10),
    (2, "2015-01-02", 25),
    (3, "2015-01-03", 20),
    (4, "2015-01-04", 30)
]

# Define schema
weather_columns = ["id", "recordDate", "temperature"]

# Create DataFrame
weather_df = spark.createDataFrame(weather_data, weather_columns)

display(weather_df)
```
